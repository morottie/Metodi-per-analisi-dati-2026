## Esercizio 3) Risoluzione di sistemi lineari

### Consegna: 

Considera le matrici

$$
A_1 = \begin{bmatrix}
10 &  2 & -1 &  1 &  0 \\
 2 &  9 &  1 & -2 &  1 \\
-1 &  1 &  7 &  1 & -1 \\
 1 & -2 &  1 &  8 &  2 \\
 0 &  1 & -1 &  2 &  6
\end{bmatrix}, \ \ \ \ 
A_2 = \begin{bmatrix} 1 & 2 \\ 2 & 1  \end{bmatrix} 
$$

e $A_3$ come matrice di Hilbert di dimensione n={2, 5, 15, 100}, poi:
* definisci il problema test $Ax=b$ per ciascuna matrice, usando come soluzione esatta il vettore di 1 di opportuna lunghezza;
* calcola il numero di condizionamento di A con `np.linalg.cond`;
* calcola la fattorizzazione LU con pivoting e quella di Cholesky (se opportuno) tramite le funzioni `lu` e `cholesky` di `scipy.linalg`;
* calcola l'errore (in norma) di fattorizzazione per A;
* risolvi il sistema lineare con le funzioni `lu_factor` e `lu_solve ` di `scipy.linalg` e calcola l'errore relativo sulla soluzione;
* quando possibile, risolvi il sistema con la fattorizzazione di Cholesky tramite la funzione `np.linalg.solve`  e calcola l'errore relativo sulla soluzione.


Definizione delle matrici

```python
import numpy as np

# Caso 1
n = 5
A1 = np.array([[10,  2, -1,  1,  0],
              [ 2,  9,  1, -2,  1],
              [-1,  1,  7,  1, -1],
              [ 1, -2,  1,  8,  2],
              [ 0,  1, -1,  2,  6]], dtype=float)

# Caso 2
#n=2
#A2 = np.array([[1, 2],
#              [2, 1]], dtype=float)

# Caso 3
#from scipy.linalg import hilbert
#n = 15          # 2 oppure 5 oppure 100
#A3 = hilbert(n)


```

Definizione del sistema lineare

```python
# Scelta della matrice
A = A1 # A1 oppure A2 oppure A3
print(f"A = {A}")

# Soluzione esatta 
x_esatta = np.ones(n)

# Calcolo del termine noto
b = A @ x_esatta
print(f"b = {b}")
```

```python
print(np.linalg.cond(A))
```

Calcolo le fattorizzazioni

```python
from scipy.linalg import lu, cholesky

P, L, U = lu(A)
print("P:\n", P)
print("L:\n", L)
print("U:\n", U)

print("Errore fattorizzazione LU:",
      np.linalg.norm(A - P @ L @ U))
```

```python
C = cholesky(A, lower=True)
print("C:\n", C)

print("Errore fattorizzazione Cholesky:",
      np.linalg.norm(A - C @ C.T))
```

Risoluzione del sistema lineare tramite LU con pivoting

```python
# Risoluzione mediante LU con pivoting
from scipy.linalg import lu_factor, lu_solve

lu, piv = lu_factor(A)
x_lu = lu_solve((lu, piv), b)
```

```python
# Per la risoluzione con LU
errore_relativo_lu = np.linalg.norm(x_lu - x_esatta) / np.linalg.norm(x_esatta)

print("Termine noto b:", b)
print("Soluzione esatta:", x_esatta)
print("Soluzione calcolata:", x_lu)
print(f"Errore relativo: {errore_relativo_lu:.3e}")
```

Risoluzione del sistema lineare tramite Cholesky

```python
# Risoluzione mediante Cholesky dei due sistemi: Ly = b e L.T x = y
y = np.linalg.solve(C, b)
x_chol = np.linalg.solve(C.T, y)
```

```python
# Per la risoluzione con Chol
errore_relativo_chol = np.linalg.norm(x_chol - x_esatta) / np.linalg.norm(x_esatta)

print("Termine noto b:", b)
print("Termine di appoggio y:", y)
print("Soluzione esatta:", x_esatta)
print("Soluzione calcolata:", x_chol)
print(f"Errore relativo: {errore_relativo_chol:.3e}")
```
