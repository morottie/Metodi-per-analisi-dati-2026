## RECAP 1: algoritmi per lo zero di una funzione



Data una funzione non lineare $f:\mathbb{R}\to\mathbb{R}$, vogliamo determinare le **radici** dell’equazione $f(x)=0$, dette anche **zeri della funzione**, cioè i valori $x^*$ tali che

$$
f(x^*)=0.
$$

A seconda della funzione, possono non esistere zeri reali, essercene uno solo, un numero finito oppure infiniti.

I **metodi numerici iterativi** consentono di approssimare uno zero costruendo una successione di valori $x_k$. Il procedimento viene ripetuto fino a soddisfare un opportuno **criterio di arresto**.


```python
import numpy as np
import matplotlib.pyplot as plt

eps = np.finfo(float).eps
print(eps)
```


### Metodo di bisezione

Richiede una funzione continua in $[a,b]$ con $f(a)\cdot f(b)<0$.

A ogni iterazione si calcola il punto medio

$$
m=\frac{a+b}{2}
$$

e si conserva la metà dell’intervallo in cui la funzione cambia segno. Se $f(m)=0$, il metodo termina.

NB:
- Garantisce la convergenza sotto le ipotesi indicate.
- Dimezza l’ampiezza dell’intervallo a ogni iterazione.
- L’errore del punto medio è al più metà dell’ampiezza dell’intervallo corrente.


![](hw2_bisezione.png)

```python
def bisezione(f, a, b, tol=1e-12, max_iter=100):

    fa = f(a)
    fb = f(b)
    iterati = []

    # Controllo degli estremi
    if fa * fb > 0:
        raise ValueError("La funzione deve cambiare segno in [a, b].")
    
    if fa == 0:
        return a, 0, iterati

    if fb == 0:
        return b, 0, iterati

    
    for k in range(1, max_iter + 1):

        # Punto medio e valore della funzione
        m = (a + b) / 2
        fm = f(m)

        # Memorizzo l'approssimazione corrente
        iterati.append(m)

        # Criterio di arresto
        if abs(b - a) / 2 <= tol  or fm == 0:
            print("Raggiunto il criterio di arresto.")
            return m, k, iterati

        # Seleziono il sottointervallo con cambio di segno, per la prossima iterazione
        if fa * fm < 0:
            b = m
        else:
            a = m
            fa = fm

    print("Raggiunto il numero massimo di iterazioni.")
    return m, k, iterati
```


### Metodo di Newton

A partire da un punto iniziale $x_0$, si calcola
$$
x_{k+1}=x_k-\frac{f(x_k)}{f'(x_k)}.
$$

Il nuovo iterato è l’intersezione della tangente al grafico in $(x_k,f(x_k))$ con l’asse delle ascisse.

NB:
- Richiede la derivata prima.
- Per uno zero semplice, con sufficiente regolarità e un punto iniziale abbastanza vicino, la convergenza è quadratica.
- Può fallire se la derivata si annulla o il punto iniziale non è adeguato.
- Per zeri multipli, la convergenza è generalmente lineare.

![](hw2_newton.png)


```python
def newton(f, df, x0, tol_x=1e-12, tol_f=1e-14, max_iter=100):

    x = x0    # x = float(x0)       
    iterati = [x]

    print(f"k = 0, x = {x:.15f}, f(x) = {f(x):.3e}")
    
    # Controllo del punto iniziale
    if abs(f(x)) <= tol_f:
        return x, 0, iterati
    
    for k in range(1, max_iter + 1):

        fx = f(x)
        dfx = df(x)

        # Controllo della derivata
        if dfx == 0:
            raise ValueError("Derivata nulla: impossibile proseguire.")

        # Nuova approssimazione
        x_nuovo = x - fx / dfx
        print(f"k = {k:2d}, x = {x_nuovo:.15f}, f(x) = {f(x_nuovo):.3e}")
        iterati.append(x_nuovo)

        # Incremento e residuo
        incremento = abs(x_nuovo - x)
        residuo = abs(f(x_nuovo))       # sarebbe: abs(f(x_nuovo) - f(x_star)) = abs(f(x_nuovo) - 0)

        # Criteri di arresto
        if incremento <= tol_x + eps * min(abs(x_nuovo), abs(x)) or residuo <= tol_f:
            print("Raggiunto un criterio di arresto.")
            return x_nuovo, k, iterati

        # Aggiornamento per la prossima iterazione
        x = x_nuovo


    print("Raggiunto il numero massimo di iterazioni.")
    return x, max_iter, iterati
```

## RECAP 2: algoritmi di discesa del gradiente per l'ottimizzazione

Data una funzione $ f : \mathbb{R}^n \to \mathbb{R}$, vogliamo risolvere il problema di minimo

$$
\min_{x \in \mathbb{R}^n} f(x).
$$

![](superficie.png)
 

### Discesa del gradiente

In questo algoritmo, si sceglie sempre la direzione di discesa come l’antigradiente, ossia 

$$ \mathbf{p}^{(k)} = -\nabla f(\mathbf{x}^{(k)}).$$

Di conseguenza, l’aggiornamento assume la forma $ \mathbf{x}^{(k+1)} = \mathbf{x}^{(k)} - \alpha_k \nabla f(\mathbf{x}^{(k)})$.

Esempio con $\alpha_k = 1$:
![Esempio con $\alpha_k = 1$](algo_GD_02.png)

#### Focus: il criterio di arresto
L’algoritmo di discesa termina quando la norma del gradiente è sufficientemente piccola

$$
\left|\nabla f(\mathbf{x}^{(k)})\right|2
\leq tol_{grad},
$$

oppure quando viene raggiunto il numero massimo di iterazioni.

```python
import numpy as np


def discesa_gradiente(
    f,              # funzione obiettivo
    grad_f,         # gradiente della funzione obiettivo
    x0,             # punto iniziale 
    alpha=1.0,      # lunghezza del passo, alpha>0
    tol_grad=1e-6,  # tolleranza sulla norma del gradiente
    maxit=1000      # massimo numero di iterazioni
):
    """
    Discesa del gradiente con passo fisso.
    """

    # Inizializzazioni
    x = np.asarray(x0, dtype=float).copy()

    k = 0
    fx = f(x)
    grad = grad_f(x)
    grad_norm = np.linalg.norm(grad)

    history = {
        "x": [x.copy()],
        "f": [fx],
        "grad_norm": [grad_norm]
    }

    # Ciclo iterativo di risoluzione
    while grad_norm > tol_grad and k < maxit:

        # Direzione di discesa
        p = -grad

        # Aggiornamento dell'iterata
        x = x + alpha * p
        k += 1

        # Aggiornamento delle quantità
        fx = f(x)
        grad = grad_f(x)
        grad_norm = np.linalg.norm(grad)

        # Salvataggio della storia
        history["x"].append(x.copy())
        history["f"].append(fx)
        history["grad_norm"].append(grad_norm)

    if grad_norm <= tol_grad:  # x è punto stazionario
        print(f"Convergenza numerica raggiunta in {k} iterazioni.")
    else:
        print("Raggiunto il numero massimo di iterazioni.")

    return x, history
```

#### Focus: il passo di discesa e il backtracking

Il parametro $\alpha_k>0$ rappresenta la lunghezza del passo di discesa. Può essere costante e non variare mai nelle iterazioni, oppure può variare usando una regola di backtracking. In questo secondo caso, si parte da un valore iniziale $\alpha_0$ e lo si riduce secondo la regola 

$$ 
\alpha_k \leftarrow \rho\alpha_k,
\qquad 0<\rho<1,
$$


finché non viene soddisfatta la condizione di Armijo:


$$
f(\mathbf{x}^{(k)} +\alpha_k \mathbf{p}^{(k)} )
\leq
f(\mathbf{x}^{(k)} ) + c_1\alpha_k \nabla f(\mathbf{x}^{(k)} )^T\mathbf{p}^{(k)} ,
\qquad 0<c_1<1.
$$

Attenzione: se $\alpha_k$ diventa troppo piccolo, il passo diventa troppo breve e non si converge velocemente alla soluzione! E' bene quindi impostare una soglia (parametro $\tau$) al di sotto del quale il backtracking si ferma per garantire $\alpha_k\geq\tau$.


Esempio con backtracking (BT):
![Esempio con bt](algo_GD_03.png)

```python
import numpy as np


def discesa_gradiente_bt(
    f,              # funzione obiettivo
    grad_f,         # gradiente della funzione obiettivo
    x0,             # punto iniziale 
    alpha0=1.0,     # lunghezza del passo iniziale, 0<alpha
    rho=0.5,        # fattore di riduzione del passo, 0<rho<1
    c1=1e-4,        # parametro della condizione di Armijo, 0<c1<1
    tau=1e-12,      # lunghezza del minimo passo ammesso, 0<rho<alpha
    tol_grad=1e-6,  # tolleranza sulla norma del gradiente
    maxit=1000      # massimo numero di iterazioni
):
    """
    Discesa del gradiente con backtracking di Armijo.
    """

    # Controlli sugli imput
    if not 0 < rho < 1:
        raise ValueError("rho deve appartenere a (0, 1).")

    if not 0 < c1 < 1:
        raise ValueError("c1 deve appartenere a (0, 1).")

    if not 0 < tau <= alpha0:
        raise ValueError("tau deve essere positivo e non superiore ad alpha0.")

    # Inizializzazioni
    x = np.asarray(x0, dtype=float).copy()

    k = 0
    fx = f(x)
    grad = grad_f(x)
    grad_norm = np.linalg.norm(grad)

    history = {
        "x": [x.copy()],
        "f": [fx],
        "grad_norm": [grad_norm],
        "alpha": []
    }

    # Ciclo iterativo di risoluzione
    while grad_norm > tol_grad and k < maxit:

        # Direzione di discesa
        p = -grad

        # Scelta del passo di discesa
        alpha = alpha0

        while f(x + alpha * p) > fx + c1 * alpha * np.dot(grad, p): # Backtracking di Armijo

            if alpha <= tau:
                print("Passo minimo raggiunto senza soddisfare Armijo.")
                return x, history

            alpha = max(rho * alpha, tau)

        # Aggiornamento dell'iterata
        x = x + alpha * p
        k += 1

        # Aggiornamento delle quantità
        fx = f(x)
        grad = grad_f(x)
        grad_norm = np.linalg.norm(grad)

        # Salvataggio della storia
        history["x"].append(x.copy())
        history["f"].append(fx)
        history["grad_norm"].append(grad_norm)
        history["alpha"].append(alpha)

    if grad_norm <= tol_grad:  # x è punto stazionario
        print(f"Convergenza numerica raggiunta in {k} iterazioni.")
    else:
        print("Raggiunto il numero massimo di iterazioni.")

    return x, history
```
